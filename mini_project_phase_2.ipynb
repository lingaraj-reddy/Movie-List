```python
df_cal=df[df['State']=='California']
```

```python
df_new=df[df['State']=='New York']
df_new.shape
```

```python
df_new.head()
```

```python
df_cal_customer = df_cal.groupby('Customer Name').agg(Sales=('Sales', 'sum'), Profit=('Profit', 'sum'))


def type_color(ex):
  if ex < 0:
    return 'red'
  else:
    return 'black'

df_cal_customer.loc[:,'blackorred'] = df_cal_customer.loc[:,'Profit'].apply(type_color)

sns.scatterplot(data=df_cal_customer, hue='blackorred' ,x='Sales',y='Profit')
plt.xlim(0,10000)
plt.ylim(-1000 ,5000)
```

```python
df_new_customer = df_new.groupby('Customer Name').agg(Sales=('Sales', 'sum'), Profit=('Profit', 'sum'))
df_new_customer.sort_values(by=['Sales'],ascending=False).head()
```
